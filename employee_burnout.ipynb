```python

# ============================================
# STANDARDIZE EMPLOYEE CODES
# ============================================

def standardize_emp_code(series):
    return (
        series.astype("string")
        .str.strip()
        .str.replace("'", "", regex=False)   # remove leading apostrophe
        .str.replace(".0", "", regex=False)  # remove Excel-style .0
        .str.lstrip("0")                    # remove leading zeros
        .replace("", pd.NA)
    )


# Create standardized employee codes
employee_master["EMP_CODE_STD"] = standardize_emp_code(
    employee_master["Employee Code"]
)

leave_application["EMP_CODE_STD"] = standardize_emp_code(
    leave_application["Employee Code"]
)

monthly_leave_balance["EMP_CODE_STD"] = standardize_emp_code(
    monthly_leave_balance["Employee Code"]
)

overtime_1["EMP_CODE_STD"] = standardize_emp_code(
    overtime_1["EMP CODE"]
)

overtime_2["EMP_CODE_STD"] = standardize_emp_code(
    overtime_2["EMP CODE"]
)

resignation["EMP_CODE_STD"] = standardize_emp_code(
    resignation["Employee Code"]
)


# ============================================
# CHECK MATCHING
# ============================================

master_set = set(
    employee_master["EMP_CODE_STD"].dropna().unique()
)

for name, df in {
    "Leave Application": leave_application,
    "Monthly Leave Balance": monthly_leave_balance,
    "OT Summary": overtime_1,
    "OT Daily": overtime_2,
    "Resignation": resignation
}.items():

    other_set = set(df["EMP_CODE_STD"].dropna().unique())

    matched = len(master_set & other_set)

    print(
        f"{name}: {matched} employees matched "
        f"({matched / len(master_set) * 100:.2f}%)"
    )
```

```python
# ============================================
# CREATE EMPLOYEE-LEVEL OT FEATURES
# ============================================

ot = overtime_2.copy()

# Make sure OT values are numeric
total_cols = [c for c in ot.columns if c.endswith("_TOT")]
chargeable_cols = [c for c in ot.columns if c.endswith("_CH")]
non_chargeable_cols = [c for c in ot.columns if c.endswith("_NCH")]

for col in total_cols + chargeable_cols + non_chargeable_cols:
    ot[col] = pd.to_numeric(ot[col], errors="coerce").fillna(0)


# Employee-level OT features
ot_features = pd.DataFrame({
    "EMP_CODE_STD": ot["EMP_CODE_STD"],

    "OT_Total_Hours": ot[total_cols].sum(axis=1),

    "OT_Avg_Daily": ot[total_cols].mean(axis=1),

    "OT_Max_Daily": ot[total_cols].max(axis=1),

    "OT_Days": (ot[total_cols] > 0).sum(axis=1),

    "OT_High_Days": (ot[total_cols] >= 2).sum(axis=1),

    "OT_Chargeable": ot[chargeable_cols].sum(axis=1),

    "OT_Non_Chargeable": ot[non_chargeable_cols].sum(axis=1)
})


print("OT feature dataset shape:", ot_features.shape)

print("\nOT feature columns:")
print(ot_features.columns.tolist())

print("\nSample:")
print(ot_features.head())

print("\nSummary:")
print(ot_features.describe())
```

```python
# ============================================
# CREATE EMPLOYEE-LEVEL LEAVE FEATURES
# ============================================

leave = leave_application.copy()

# Clean numeric leave days
leave["Actual Leave Days"] = pd.to_numeric(
    leave["Actual Leave Days"],
    errors="coerce"
).fillna(0)

# Define leave status groups
approved_status = [
    "Approved by RM",
    "Auto Approved"
]

pending_status = [
    "Submitted"
]

# Flags
leave["Is_Approved"] = leave["Leave Status"].isin(
    approved_status
)

leave["Is_Pending"] = leave["Leave Status"].isin(
    pending_status
)


# --------------------------------------------
# Aggregate by employee
# --------------------------------------------

leave_features = (
    leave.groupby("EMP_CODE_STD")
    .agg(
        Leave_Total_Days=(
            "Actual Leave Days",
            lambda x: x[leave.loc[x.index, "Is_Approved"]].sum()
        ),

        Leave_Applications=(
            "Is_Approved",
            "sum"
        ),

        Leave_Avg_Days=(
            "Actual Leave Days",
            lambda x: (
                x[leave.loc[x.index, "Is_Approved"]].mean()
                if leave.loc[x.index, "Is_Approved"].any()
                else 0
            )
        ),

        Leave_Max_Days=(
            "Actual Leave Days",
            lambda x: (
                x[leave.loc[x.index, "Is_Approved"]].max()
                if leave.loc[x.index, "Is_Approved"].any()
                else 0
            )
        ),

        Leave_Pending_Applications=(
            "Is_Pending",
            "sum"
        )
    )
    .reset_index()
)


print("Leave feature dataset shape:", leave_features.shape)

print("\nColumns:")
print(leave_features.columns.tolist())

print("\nSample:")
print(leave_features.head())

print("\nSummary:")
print(leave_features.describe())
```

```python
# ============================================================
# STEP 1: CREATE OT DAILY TREND FEATURES
# ============================================================

import pandas as pd
import numpy as np

# Make a copy of OT Daily data
ot_daily = overtime_2.copy()

# Standardize employee code
ot_daily["EMP_CODE_STD"] = pd.to_numeric(
    ot_daily["EMP CODE"],
    errors="coerce"
)

# ------------------------------------------------------------
# Find all daily TOTAL OT columns
# Example: 01-JUN-26_TOT, 02-JUN-26_TOT, ...
# ------------------------------------------------------------

daily_total_cols = [
    col for col in ot_daily.columns
    if str(col).upper().endswith("_TOT")
]

print("Number of daily OT columns:", len(daily_total_cols))
print("Daily OT columns:")
print(daily_total_cols)


# Convert daily OT columns to numeric
for col in daily_total_cols:
    ot_daily[col] = pd.to_numeric(
        ot_daily[col],
        errors="coerce"
    ).fillna(0)


# ------------------------------------------------------------
# Create daily OT matrix
# ------------------------------------------------------------

daily_ot_matrix = ot_daily[daily_total_cols].copy()

# Total OT during the month
ot_daily["OT_Total_Hours"] = daily_ot_matrix.sum(axis=1)

# Number of days with OT
ot_daily["OT_Days"] = (daily_ot_matrix > 0).sum(axis=1)

# Maximum OT in one day
ot_daily["OT_Max_Daily"] = daily_ot_matrix.max(axis=1)

# Average OT across all calendar days
ot_daily["OT_Avg_Daily"] = daily_ot_matrix.mean(axis=1)

# Number of high-OT days
# Using >= 6 hours as high OT day
ot_daily["OT_High_Days"] = (daily_ot_matrix >= 6).sum(axis=1)


print("\nOT daily features created.")
print(
    ot_daily[
        [
            "EMP_CODE_STD",
            "OT_Total_Hours",
            "OT_Avg_Daily",
            "OT_Max_Daily",
            "OT_Days",
            "OT_High_Days"
        ]
    ].head()
)
```

```python
# ============================================================
# STEP 2: ROLLING + LAGGED OT FEATURES
# ============================================================

# Create a long-format daily dataset

ot_long = ot_daily[
    ["EMP_CODE_STD"] + daily_total_cols
].melt(
    id_vars="EMP_CODE_STD",
    value_vars=daily_total_cols,
    var_name="Date_Column",
    value_name="OT_Hours"
)

# Extract date from column names
# Example: 01-JUN-26_TOT -> 01-JUN-26

ot_long["Date"] = (
    ot_long["Date_Column"]
    .str.replace("_TOT", "", regex=False)
)

ot_long["Date"] = pd.to_datetime(
    ot_long["Date"],
    format="%d-%b-%y",
    errors="coerce"
)

ot_long["OT_Hours"] = pd.to_numeric(
    ot_long["OT_Hours"],
    errors="coerce"
).fillna(0)

# Sort properly
ot_long = ot_long.sort_values(
    ["EMP_CODE_STD", "Date"]
)

print("Long OT dataset shape:", ot_long.shape)
print(ot_long.head())

```

```python

# ============================================================
# REBUILD OT LONG DATA + TREND FEATURES
# ============================================================

import pandas as pd
import numpy as np

# ------------------------------------------------------------
# STEP 1: Identify daily OT total columns
# ------------------------------------------------------------

ot_total_cols = [
    col for col in overtime_2.columns
    if col.endswith("_TOT") and "-" in col
]

print("Number of OT daily columns:", len(ot_total_cols))
print("First 5:", ot_total_cols[:5])
print("Last 5:", ot_total_cols[-5:])


# ------------------------------------------------------------
# STEP 2: Keep only one row per employee
# ------------------------------------------------------------

ot_source = overtime_2[
    ["EMP_CODE_STD"] + ot_total_cols
].copy()

print("\nOT rows before removing duplicates:", len(ot_source))
print(
    "Unique employees before:",
    ot_source["EMP_CODE_STD"].nunique()
)

# Your earlier data showed 4 duplicate employee rows
ot_source = ot_source.drop_duplicates(
    subset=["EMP_CODE_STD"],
    keep="first"
).copy()

print(
    "OT rows after removing duplicates:",
    len(ot_source)
)
print(
    "Unique employees after:",
    ot_source["EMP_CODE_STD"].nunique()
)


# ------------------------------------------------------------
# STEP 3: Convert wide OT data to long format
# ------------------------------------------------------------

ot_long = ot_source.melt(
    id_vars=["EMP_CODE_STD"],
    value_vars=ot_total_cols,
    var_name="Date_Column",
    value_name="OT_Hours"
)

# Extract date from column name
# Example: 01-JUN-26_TOT -> 01-JUN-26
ot_long["Date"] = pd.to_datetime(
    ot_long["Date_Column"].str.replace("_TOT", "", regex=False),
    format="%d-%b-%y",
    errors="coerce"
)

# Convert OT hours to numeric
ot_long["OT_Hours"] = pd.to_numeric(
    ot_long["OT_Hours"],
    errors="coerce"
).fillna(0)

# Sort properly before calculating lag/rolling features
ot_long = ot_long.sort_values(
    ["EMP_CODE_STD", "Date"]
).reset_index(drop=True)


# ------------------------------------------------------------
# STEP 4: Previous day's OT
# ------------------------------------------------------------

ot_long["OT_Lag_1_Day"] = (
    ot_long
    .groupby("EMP_CODE_STD")["OT_Hours"]
    .shift(1)
    .fillna(0)
)


# ------------------------------------------------------------
# STEP 5: Previous 7-day average
# ------------------------------------------------------------

ot_long["OT_Rolling_7D"] = (
    ot_long
    .groupby("EMP_CODE_STD")["OT_Hours"]
    .transform(
        lambda x:
        x.shift(1)
         .rolling(window=7, min_periods=1)
         .mean()
    )
    .fillna(0)
)


# ------------------------------------------------------------
# STEP 6: Previous 14-day average
# ------------------------------------------------------------

ot_long["OT_Rolling_14D"] = (
    ot_long
    .groupby("EMP_CODE_STD")["OT_Hours"]
    .transform(
        lambda x:
        x.shift(1)
         .rolling(window=14, min_periods=1)
         .mean()
    )
    .fillna(0)
)


# ------------------------------------------------------------
# STEP 7: Final check
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("OT LONG DATA")
print("=" * 60)

print("Shape:", ot_long.shape)

print("\nColumns:")
print(ot_long.columns.tolist())

print("\nSample:")
print(
    ot_long[
        [
            "EMP_CODE_STD",
            "Date",
            "OT_Hours",
            "OT_Lag_1_Day",
            "OT_Rolling_7D",
            "OT_Rolling_14D"
        ]
    ].head(15)
)

print("\nUnique employees:")
print(ot_long["EMP_CODE_STD"].nunique())

print("\nDuplicate employee-date combinations:")
print(
    ot_long.duplicated(
        ["EMP_CODE_STD", "Date"]
    ).sum()
)
```

```python
# ============================================================
# STEP 4: CREATE EMPLOYEE-LEVEL OT TREND FEATURES
# For Autoencoder Burnout Model
# ============================================================

# Aggregate daily OT behaviour to one row per employee

ot_trend_features = (
    ot_long
    .groupby("EMP_CODE_STD")
    .agg(
        OT_Mean_Daily=("OT_Hours", "mean"),
        OT_Max_Daily=("OT_Hours", "max"),
        OT_Total_Days=("OT_Hours", lambda x: (x > 0).sum()),
        OT_High_Days=("OT_Hours", lambda x: (x >= 4).sum()),

        # Recent/latest workload indicators
        OT_Latest_Day=("OT_Hours", "last"),
        OT_Latest_Lag_1D=("OT_Lag_1_Day", "last"),
        OT_Latest_Rolling_7D=("OT_Rolling_7D", "last"),
        OT_Latest_Rolling_14D=("OT_Rolling_14D", "last"),

        # Average historical rolling workload
        OT_Avg_Rolling_7D=("OT_Rolling_7D", "mean"),
        OT_Avg_Rolling_14D=("OT_Rolling_14D", "mean"),

        # Maximum sustained workload
        OT_Max_Rolling_7D=("OT_Rolling_7D", "max"),
        OT_Max_Rolling_14D=("OT_Rolling_14D", "max")
    )
    .reset_index()
)


# ------------------------------------------------------------
# CHECK
# ------------------------------------------------------------

print("=" * 60)
print("EMPLOYEE-LEVEL OT TREND FEATURES")
print("=" * 60)

print("\nShape:")
print(ot_trend_features.shape)

print("\nColumns:")
print(ot_trend_features.columns.tolist())

print("\nUnique employees:")
print(
    ot_trend_features["EMP_CODE_STD"].nunique()
)

print("\nDuplicate employee codes:")
print(
    ot_trend_features["EMP_CODE_STD"].duplicated().sum()
)

print("\nSample:")
print(
    ot_trend_features.head()
)

print("\nSummary:")
print(
    ot_trend_features.describe()
)
```

```python

# ============================================================
# STEP 10: PREPARE AUTOENCODER INPUT
# ============================================================

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline

# ------------------------------------------------------------
# Numeric features
# ------------------------------------------------------------

numeric_cols = [
    "OT_Mean_Daily",
    "OT_Max_Daily",
    "OT_Total_Days",
    "OT_High_Days",
    "OT_Latest_Day",
    "OT_Latest_Lag_1D",
    "OT_Latest_Rolling_7D",
    "OT_Latest_Rolling_14D",
    "OT_Avg_Rolling_7D",
    "OT_Avg_Rolling_14D",
    "OT_Max_Rolling_7D",
    "OT_Max_Rolling_14D",
    "Leave_Total_Days",
    "Leave_Applications",
    "Leave_Avg_Days",
    "Leave_Max_Days",
    "Leave_Pending_Applications",
    "Year Of Service (Current DOJ)"
]

# ------------------------------------------------------------
# Low-cardinality categorical features
# ------------------------------------------------------------

categorical_cols = [
    "Department",
    "Grade",
    "Band",
    "Employee Category",
    "Branch"
]

print("Numeric features:", len(numeric_cols))
print("Categorical features:", len(categorical_cols))


# ------------------------------------------------------------
# Preprocessing
# ------------------------------------------------------------

numeric_pipeline = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "scaler",
            StandardScaler()
        )
    ]
)

categorical_pipeline = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="most_frequent")
        ),
        (
            "onehot",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            )
        )
    ]
)


preprocessor = ColumnTransformer(
    transformers=[
        (
            "numeric",
            numeric_pipeline,
            numeric_cols
        ),
        (
            "categorical",
            categorical_pipeline,
            categorical_cols
        )
    ]
)


# ------------------------------------------------------------
# Create feature matrix
# ------------------------------------------------------------

X_raw = burnout_model_data[
    numeric_cols + categorical_cols
].copy()

X_processed = preprocessor.fit_transform(X_raw)

print("\n========================================")
print("AUTOENCODER INPUT")
print("========================================")

print("Original rows:", X_raw.shape[0])
print("Processed shape:", X_processed.shape)
print("Data type:", X_processed.dtype)
```

```python

# ============================================================
# STEP 11: AUTOENCODER NEURAL NETWORK
# ============================================================

import numpy as np
import tensorflow as tf
from tensorflow.keras import Model
from tensorflow.keras.layers import Input, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping

# Reproducibility
np.random.seed(42)
tf.random.set_seed(42)

# ------------------------------------------------------------
# INPUT
# ------------------------------------------------------------

X = np.asarray(X_processed, dtype=np.float32)

input_dim = X.shape[1]

print("=" * 60)
print("AUTOENCODER MODEL")
print("=" * 60)
print("Input shape:", X.shape)
print("Input features:", input_dim)


# ------------------------------------------------------------
# AUTOENCODER ARCHITECTURE
# ------------------------------------------------------------

input_layer = Input(shape=(input_dim,))

# Encoder
x = Dense(32, activation="relu")(input_layer)
x = Dropout(0.10)(x)

x = Dense(16, activation="relu")(x)

# Bottleneck
bottleneck = Dense(
    8,
    activation="relu",
    name="bottleneck"
)(x)

# Decoder
x = Dense(16, activation="relu")(bottleneck)
x = Dropout(0.10)(x)

x = Dense(32, activation="relu")(x)

# Reconstruction
output_layer = Dense(
    input_dim,
    activation="linear"
)(x)


# ------------------------------------------------------------
# BUILD MODEL
# ------------------------------------------------------------

autoencoder = Model(
    inputs=input_layer,
    outputs=output_layer
)

autoencoder.compile(
    optimizer="adam",
    loss="mse"
)

print("\nModel architecture:")
autoencoder.summary()


# ------------------------------------------------------------
# TRAINING
# ------------------------------------------------------------

early_stopping = EarlyStopping(
    monitor="val_loss",
    patience=10,
    restore_best_weights=True
)

print("\nStarting training...")

history = autoencoder.fit(
    X,
    X,
    epochs=100,
    batch_size=256,
    validation_split=0.20,
    shuffle=True,
    callbacks=[early_stopping],
    verbose=1
)


# ------------------------------------------------------------
# TRAINING RESULT
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("AUTOENCODER TRAINING COMPLETE")
print("=" * 60)

print(
    "Epochs completed:",
    len(history.history["loss"])
)

print(
    "Final training loss:",
    history.history["loss"][-1]
)

print(
    "Final validation loss:",
    history.history["val_loss"][-1]
)
```

```python
# ============================================================
# STEP 12: CALCULATE AUTOENCODER RECONSTRUCTION ERROR
# ============================================================

# Reconstruct the input data
X_reconstructed = autoencoder.predict(
    X,
    verbose=0
)

print("=" * 60)
print("RECONSTRUCTION")
print("=" * 60)

print("Original shape:      ", X.shape)
print("Reconstructed shape: ", X_reconstructed.shape)


# ------------------------------------------------------------
# RECONSTRUCTION ERROR FOR EACH EMPLOYEE
# ------------------------------------------------------------

reconstruction_error = np.mean(
    np.square(X - X_reconstructed),
    axis=1
)

print("\nReconstruction error calculated.")
print("Number of employees:", len(reconstruction_error))


# ------------------------------------------------------------
# ADD SCORE TO BURNOUT DATASET
# ------------------------------------------------------------

burnout_scored = burnout_model_data.copy()

burnout_scored["Burnout_Score"] = reconstruction_error


# ------------------------------------------------------------
# BASIC SCORE CHECK
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("BURNOUT SCORE SUMMARY")
print("=" * 60)

print(
    burnout_scored["Burnout_Score"].describe()
)

print("\nTop 10 highest burnout scores:")

print(
    burnout_scored[
        ["EMP_CODE_STD", "Burnout_Score"]
    ]
    .sort_values(
        "Burnout_Score",
        ascending=False
    )
    .head(10)
)
```

```python

# ============================================================
# STEP 15: BUSINESS-LEVEL BURNOUT FEATURE IMPORTANCE
# ============================================================

importance_df = global_importance.copy()

# Original categorical features used in the burnout model
categorical_cols = [
    "Department",
    "Designation",
    "Grade",
    "Band",
    "Employee Category",
    "Branch"
]


def get_original_feature(processed_name):

    # Numeric features
    if processed_name.startswith("numeric__"):
        return processed_name.replace("numeric__", "", 1)

    # One-hot encoded categorical features
    if processed_name.startswith("categorical__"):

        name = processed_name.replace("categorical__", "", 1)

        # Find which original categorical column it belongs to
        for col in categorical_cols:
            if name == col or name.startswith(col + "_"):
                return col

        return name

    return processed_name


# Map processed features back to original business features
importance_df["Original_Feature"] = (
    importance_df["Processed_Feature"]
    .apply(get_original_feature)
)


# Aggregate reconstruction error
business_importance = (
    importance_df
    .groupby("Original_Feature", as_index=False)
    ["Mean_Reconstruction_Error"]
    .sum()
    .sort_values(
        "Mean_Reconstruction_Error",
        ascending=False
    )
    .reset_index(drop=True)
)


# Calculate percentage contribution
total_error = business_importance[
    "Mean_Reconstruction_Error"
].sum()

business_importance["Importance_%"] = (
    business_importance["Mean_Reconstruction_Error"]
    / total_error
    * 100
)


print("=" * 65)
print("BUSINESS-LEVEL BURNOUT FEATURE IMPORTANCE")
print("=" * 65)

print(
    business_importance.to_string(index=False)
)


print("\n" + "=" * 65)
print("TOP 10 BURNOUT DRIVERS")
print("=" * 65)

print(
    business_importance
    .head(10)
    .to_string(index=False)
)
```
